##2) Crear el chatbot utilizando TFIDF y similitud del coseno

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

#prompt: Separar las preguntas y respuestas del diccionario
preguntas = list(datos.keys())
respuestas = list(datos.values())

#Vectorización con TF-IDF
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(preguntas)

def chatbot(consulta_usuario):
  consulta_vec = vectorizer.transform([consulta_usuario])
  similitudes = cosine_similarity(consulta_vec, tfidf_matrix).flatten()

  indice_mejor_coincidencia = similitudes.argmax()

  return respuestas[indice_mejor_coincidencia]


while True:
  entrada_usuario = input("Tú: " )
  if entrada_usuario.lower() == "chau":
    print("Chatbot: ¡Hasta luego!")
    break

  respuesta_chatbot, similitud_maxima = chatbot(entrada_usuario)
  print("Chatbot:", respuesta_chatbot)
```

##Explicación del código:

Primero creé un diccionario con las preguntas frecuentes que los clientes suelen realizar y sus respectivas respuestas.

Luego utilicé `TfidfVectorizer` para transformar las preguntas en representaciones numéricas, según lo indicaba el enunciado. TF-IDF le asigna mayor importancia a las palabras que son más relevantes dentro del conjunto de preguntas.

Usé `cosine_similarity` para comparar la consulta del cliente con las preguntas en el dataset, y determinar cuál tiene mayor similitud con la misma, para luego asignarle la respuesta correspondiente.

A continuación creé una función que toma la consulta del cliente, la vectoriza y calcula la similitud. Con `argmax()` se identifica el índice de la pregunta con mayor similitud.

Para la interacción con el cliente implementé un bucle while igual al que vimos en clase en un ejercicio, que permite realizar múltiples consultas, y que finaliza cuando se escribe la palabra "chau".

```python
import spacy
import numpy as np

#Cargo el modelo en español preentrenado
nlp = spacy.load("es_core_news_md")

#prompt: Obtener los embeddings de las preguntas
preguntas = list(datos.keys())
respuestas = list(datos.values())

embeddings_preguntas = [nlp(pregunta).vector for pregunta in preguntas]
```

```python
#prompt: Crear un chatbot utilizando embeddings y similitud del coseno
from sklearn.metrics.pairwise import cosine_similarity

def chatbot_embeddings(consulta_usuario):
  consulta_vec = nlp(consulta_usuario).vector.reshape(1, -1)
  similitudes = cosine_similarity(consulta_vec, embeddings_preguntas).flatten()

  indice_mejor_coincidencia = similitudes.argmax()

  return respuestas[indice_mejor_coincidencia]


while True:
  entrada_usuario = input("Tú: " )
  if entrada_usuario.lower() == "chau":
    print("Chatbot: ¡Hasta luego!")
    break

  respuesta_chatbot = chatbot_embeddings(entrada_usuario)
  print("Chatbot:", respuesta_chatbot)

```

##Explicación del código:

El código de este chatbot es similar al del ejercicio anterior, ya que se utiliza el mismo diccionario de preguntas y respuestas y la similitud del coseno para encontrar la mejor coincidencia.

La diferencia está en la forma de vectorizar las preguntas, ya que en este caso usé el modelo preentrenado `es_core_news_md` de spaCy.

Con `nlp(pregunta).vector` obtengo el embedding de cada pregunta del dataset, trasnformándola en un vector numérico que realiza las relaciones semánticas entre las palabras.

Cuando el cliente realiza una consulta también se transforma en un embedding y se compara con loos vectores de las preguntas almacenadas mediante la similitud del coseno. Así es como el chatbot identifica similitudes entre consultas aunque no se usen exactamente las mismas palabras.

##Justificación del embedding elegido:

Elegí el modelo preentrenado `es_core_news_md` de la librería spaCy principalmente porque procesa textos en español, y las consultas que realizan nuestros clientes son todas en español.

Además este modelo puede asociar las consultas aunque se usen diferentes palabras o expresiones, ya que trabaja con vectores que capturan las relaciones semánticas. Es decir, reconoce distintas formas de realizar una misma pregunta

TF-IDF: trabaja con las palabras que se encuentran en nuestro conjunto de palabras, y como el diccionario es pequeño, identifica mejor las palabras clave de cada pregunta.

Embeddings: trabaja con vectores de 300 dimensiones y representaciones aprendidas a partir de grandes cantidades de texto, es por eso que al ser el diccionario tan pequeño es como se "se pierde".

##**Conclusión general**

Conclusión general

A partir de las pruebas realizadas, pude observar que el chatbot basado en TF-IDF obtuvo mejores resultados que el de embeddings, respondiendo correctamente el 83,3 % de las consultas, frente al 66,7 % de embeddings.

Si bien esperaba que embeddings tuviera un mejor rendimiento, los resultados demostraron que un modelo más complejo no necesariamente funciona mejor en todos los casos.

Considero que TF-IDF resultó más efectivo porque trabajamos con un diccionario pequeño de preguntas, donde identificar las palabras clave fue suficiente para seleccionar correctamente la mayoría de las respuestas. Sin embargo, presentó dificultades cuando las palabras estaban expresadas de manera diferente, por ejemplo, en singular y plural. Por otro lado, embeddings logró reconocer algunas relaciones semánticas que TF-IDF no pudo identificar, pero también confundió preguntas que requerían respuestas completamente diferentes.

Una buena forma de encontrarle una solución a esto es agregando un umbral de similitudes, es decir, descartar aquellas similitudes que tengan un valor bajo, y que el chatbot envíe un mensaje, que podría ser: "Disculpe, no tengo información suficiente para responder esa pregunta. Si quiere, puede intentar reformulándola". Así, el cliente escribiría la misma pregunta con otras palabras y el chatbot tendría una oportunidad más de encontrar la respuesta correcta. O sino, otra opción sería que el chatbot simplemente responda "Disculpe, no tengo información suficiente para responder esa pregunta" y el tema se cierra ahí, entonces el cliente deberá formular una pregunta de otro tema.

En conclusión, para este chatbot de atención al cliente de Giulya, TF-IDF resultó ser la alternativa más efectiva. Esto no significa que siempre sea superior a embeddings, ya que el rendimiento depende del conjunto de datos, del modelo utilizado y del tipo de consultas.
